```python
import numpy as np
from scipy import constants
from wave_optics_propagation.visualization import plot_wavefunction
from wave_optics_propagation_enhanced.classical_numerics import (
    free_space_signal_generator,
)
from wave_optics_propagation_enhanced.parameters import ExperimentParameters
```

```python
# scale all lengths down by 1e-3 for better numerical stability
scaling = 1.493

experiment_params = ExperimentParameters(
    vacuum_wavelength=1e-6,
    beam_FWHM=20e-3 * scaling,
    focal_length=200e-3 * scaling,
    refractive_index=1.25,
    propagation_after_lens=1.5 * 200e-3 * scaling,
    transverse_length=100e-3 * scaling,
    num_of_steps_after_lens=300,
    lens_slices=10000,
    num_qubits=6,
    max_delta=0.01,
    lens_reverse_order=True,
    fresnel_approximation=True,
    scale_down_phases=True,
)
```

```python
index = 3

plot_wavefunction(experiment_params.lens_signals[index].data, normalize=False)

plot_wavefunction(
    np.fft.fft(experiment_params.lens_signals[index].data, norm="ortho"),
    normalize=False,
)

phase = np.exp(1j * experiment_params.lens_signals[index].data)
plot_wavefunction(phase, normalize=False)
plot_wavefunction(np.fft.fft(phase, norm="ortho"), normalize=False)
```

```python
np.max(experiment_params.lens_signals[index].data)
```

```python
factor = (
    (experiment_params.refractive_index - 1)
    / experiment_params.vacuum_wavelength
    * experiment_params.lens_slice_thickness
)
factor
```

```python
1 / factor
```

```python
propagator_signal = free_space_signal_generator(
    k_axis=experiment_params.k_axis,
    delta_t=experiment_params.lens_slice_thickness / constants.c,
    wavelength=experiment_params.vacuum_wavelength,
    c=constants.c,
)

plot_wavefunction(propagator_signal.data, normalize=False)
plot_wavefunction(
    np.fft.fft(propagator_signal.data, norm="ortho"),
    normalize=False,
)

phase = np.exp(1j * propagator_signal.data)
plot_wavefunction(phase, normalize=False)
plot_wavefunction(np.fft.fft(phase, norm="ortho"), normalize=False)
```
