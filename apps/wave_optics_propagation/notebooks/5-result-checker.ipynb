```python
import os

path = "../../../.result/3-lens-simulation-tuning-before-decoupling"

simulations = [f for f in os.listdir(path) if os.path.isdir(os.path.join(path, f))]
```

```python
import pandas as pd
from wave_optics_propagation.storage import load_initial_parameters

df = pd.DataFrame()
for sim in simulations:
    sim_path = os.path.join(path, sim)
    initial_params = load_initial_parameters(sim_path)
    initial_params["path_to_simulation"] = sim_path
    df = pd.concat(
        [df, pd.DataFrame([initial_params])],
        ignore_index=True,
    )
    # df.loc[len(df)] = initial_params
```
